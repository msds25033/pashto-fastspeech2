## Step 1: phone sequence -> audio
Phones separated by spaces; `sp` is a pause.

```python
PHONES = "{i z a t m a n d o m j ɑ n d o sp x u j ɑ n d o m a ʃ r ɑ n o sp k a ʃ r ɑ n o sp d a r a b j a l a w ə l d a sp m j ɑ ʃ t e sp p a s ə r l e sp m ɑ ʃ ɑ sp a l ɑ sp r a w ɑ n d e sp d a h u z u r p ɑ k s a l i a l ɑ a l a j h sp w a s l a m sp t ɑ r i f a l ɑ sp p ɑ k p h i sp q u r ɑ n p ɑ k sp k e sp k ɽ e d e}"

Audio(tts.speak_phones(PHONES), rate=tts.SR)
```

## Step 2: Pashto text -> phone sequence
Words missing from the lexicon are skipped with a warning; punctuation becomes a pause (`sp`).

```python
TEXT = "دَ ربیع الاول دَ میاشتې پسرلې ماشاء الله روان دے۔"

print(tts.text_to_phones(TEXT))
```

## Step 3: Pashto text -> audio (via the phones)

```python
TEXT = "دَ ربیع الاول دَ میاشتې پسرلې ماشاء الله روان دے۔"

Audio(tts.speak_text(TEXT), rate=tts.SR)
```
